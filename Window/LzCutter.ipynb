```python
# Given values
a = 1  # Absorptivity
Q = 1000  # Specific energy (J/g)
rho = 9.7 * 10**(-4)  # Density (g/mm³)
R = 0.25  # Beam radius (mm)
P_max = 30  # Max lazer power (W) 
V = 25.4  # Cutting speed (mm/s)
D = 0.8 # Thickness (mm)
```

```python
#D = a * (P / V)/ (Q * rho * R * np.sqrt(np.pi)) 
```

```python
thickness = np.linspace(0,1,100)
P_perc = ((Q * rho * R * np.sqrt(np.pi) * V * thickness))/P_max
```

```python
(Q * rho * R * np.sqrt(np.pi) * V * 2)/P_max
```

```python
plt.plot(thickness,P_perc*100)
plt.xlabel("Thickness (mm)")
plt.ylabel("Percent Power %" )
```

```python
Vel = np.linspace(0,25.4,100)
D = a * (P_max*0.5 / Vel)/ (Q * rho * R * np.sqrt(np.pi))

plt.plot(Vel/25.4*100,D)
plt.xlabel("Percent Speed (%)")
plt.ylabel("Thickness (mm)" )
```

```python
# Define P and V ranges
POWER = np.linspace(0.01, 30, 50)  # Avoiding division by zero
VELOCITY = np.linspace(0.01, 25.4, 50)
POWER, VELOCITY = np.meshgrid(POWER, VELOCITY)

# Compute D
DEPTH = a * (POWER / VELOCITY) / (Q * rho * R * np.sqrt(np.pi))

# Plotting
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(POWER, VELOCITY, DEPTH, cmap='viridis')

# Labels
ax.set_xlabel('Power')
ax.set_ylabel('Velocity')
ax.set_zlabel('Depth')
ax.set_title('3D Plot of D = a * (P / V) / (Q * rho * R * sqrt(pi))')

plt.show()
```
