```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import re
```

```python
df = pd.read_csv("/Users/thachdang/TIM_Hardware/Level_meter/Level Meter - Combined.csv")
```

```python
df.head()
```

```python
#Height of level meter reading
H_60 = df['Height (mm) 60 deg']
H_45 = df["Height (mm) 45 deg"]
H_30 = df['Height (mm) 30 deg']
H_15 = df["Height (mm) 15 deg"]
H_0 = df['Height (mm) 0 deg']
```

```python
#Volume of LHe 
V_60 = df['Volume (L) 60 deg']
V_45 = df['Volume (L) 45 deg']
V_30 = df['Volume (L) 30 deg']
V_15 = df['Volume (L) 15 deg']
V_0 = df['Volume (L) 0 deg'] 
```

```python
plt.figure(figsize=(15, 10))

plt.plot(H_60/10,V_60, label = '60 degrees tilt', linestyle='--', color='blue', marker='x')
plt.plot(H_45/10,V_45, label = '45 degrees tilt',  linestyle='-', color='orange', marker='o')
plt.plot(H_30/10,V_30, label = '30 degrees tilt',  linestyle='-.', color='g', marker='s')
plt.plot(H_15/10,V_15, label = '15 degrees tilt',  linestyle=':', color='r', marker='^')
plt.plot(H_0/10,V_0, label = '0 degrees tilt',  marker='.', color='purple')
plt.xlabel('Height (cm)')
plt.ylabel('Volume (L)')
plt.xlim(0,int(max(H_45/10))+1)
plt.ylim(0,max(V_45)+5)
plt.legend(loc = 'upper left')

plt.xticks(np.arange(0, int(max(H_45/10))+1, 1))
plt.yticks(np.arange(0, int(max(V_45))+1, 5))
plt.title("TIM Cryostat Helium Level")
plt.grid()
plt.show()

plt.savefig('level_meter_reading.png', dpi=300, bbox_inches='tight')
```
